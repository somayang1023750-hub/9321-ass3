## Dataset Description

Each record represents one movie. The tasks are to predict movie revenue and movie rating.  
Below is a summary of the key columns:

| Column | Description | Notes |
|--------|-------------|-------|
| imdb_id | Unique IMDb identifier for each movie | Identifier |
| title | Movie title | Text |
| vote_average | Average user rating of the movie | Numerical; 0–10 |
| vote_count | Number of user votes contributing to the movie rating | Numerical |
| status | Release status of the movie | Categorical; all values are Released |
| release_date | Official movie release date | Date; YYYY-MM-DD; contains missing values |
| revenue | Total reported box-office revenue of the movie | Numerical; **Regression target** |
| runtime | Duration of the movie | Numerical; minutes |
| budget | Production budget of the movie | Numerical; contains zero values |
| original_language | Original language of the movie | Categorical; language code |
| original_title | Movie title in its original language | Text |
| overview | Plot summary or description of the movie | Text; contains missing values |
| popularity | Movie popularity score | Numerical |
| tagline | Promotional slogan associated with the movie | Text; contains missing values |
| genres | Movie genres | Multiple comma-separated values; contains missing values |
| production_companies | Companies involved in movie production | Multiple comma-separated values; contains missing values |
| production_countries | Countries where the movie was produced | Multiple comma-separated values; contains missing values |
| spoken_languages | Languages spoken in the movie | Multiple comma-separated values; contains missing values |
| cast | Actors and actresses appearing in the movie | Multiple comma-separated values; contains missing values |
| director | Director of the movie | Text; contains missing values |
| director_of_photography | Person responsible for cinematography | Text; contains missing values |
| writers | Writers or screenwriters of the movie | Multiple comma-separated values; contains missing values |
| producers | Producers involved in the movie | Multiple comma-separated values; contains missing values |
| music_composer | Composer responsible for the movie's music | Text; contains missing values |
| rating | Categorised movie rating | Low, Medium, High; **Classification target** |
| imdb_votes | Number of votes received on IMDb | Numerical |
| keywords | Descriptive keywords associated with the movie | Multiple pipe-separated values; contains missing values |

## Important Rules and Penalties

1. **Dataset Usage Rules**
   - You may use the **training dataset** for model training and validation.  
   - The supplied `test.csv` includes the true `revenue` and `rating` values for evaluation only. You may use these target values to calculate evaluation metrics for your predictions. Do not use them as model inputs or for training, feature selection, preprocessing, or hyperparameter tuning.  
   - Your model will also be evaluated against **a different test dataset** (only available to tutors); the training dataset will remain the same during marking. The dataset schema will be the same for all datasets (e.g. same structure, similar quality, similar class balance). However, you should always cater for unexpected values—for example, when using a label encoder on a categorical variable, you may encounter previously unseen values. 

2. **Target Leakage**
   - Avoid any form of target leakage.  
     - Do **not** engineer features that indirectly encode target information.  
     - You cannot use `rating` as a feature for predicting `revenue`, and you cannot use `revenue` as a feature for predicting `rating`.  

3. **Late Submission Penalty**
   - 5% deduction per day (out of total marks).  
   - Submissions after **day 5** will **not be marked**.

4. **Runtime**
   - Evaluation will be conducted on the CSE servers using `Python 3.13` and the packages specified in [requirements.txt](requirements.txt).
   - Use the provided [requirements.txt](requirements.txt) to set up your environment.
   - Use only the libraries available in that environment (e.g., `pandas`, `scikit-learn`, `imbalanced-learn`, etc.).
      - The use of other external libraries, packages, or tools is not permitted, unless explicitly approved by the instructors. 
   - If your models take **more than 2 minutes in total** to train and generate both outputs on CSE lab machines, you will lose **1 mark per additional minute or part thereof**.

5. **Automatic Zero (0) Mark Conditions**  
   You will receive a **zero mark** for Assignment 3 if:
   - Your code produces **hard-coded predictions** instead of using a trained model.  
   - You use the **test dataset for training or oversampling**.

Your code must read files using only `sys.argv[1]` and `sys.argv[2]` (or equivalent argument parsing).
   - You **do not produce the required output files properly** (`z{studentID}_regression.csv`, `z{studentID}_classification.csv`). Make sure your output files are named correctly and formatted as expected with the correct column names and the number of rows must be the same as that of `<test_file>`.

**Part II: Regression (5 marks)** 

Here is how your mark will be calculated:

$$
\text{regression}_{\text{mark}} =
\begin{cases} 
0 & \text{if } RMSLE \ge 2.2 \text{ or invalid output files (size and format) }  \\[1mm] 
5 & \text{if } RMSLE \le 1.7 \\[1mm]
(1- \frac{RMSLE - 1.7}{2.2 - 1.7}) \times 5 & \text{if } 1.7 < RMSLE < 2.2
\end{cases}
$$

**RMSLE**:root_mean_squared_log_error  will be calculated using your `z{studentID}_regression.csv` output file.


The requirement may be relaxed — awarding full marks for root_mean_squared_log_error $\le$ 1.7 — before releasing the marks for Assignment 3, if no student meets the original threshold.

**Part III: Classification (5 marks)** 

Here is how your mark will be calculated:

$$
\text{classification}_{\text{mark}} =
\begin{cases} 
0 & \text{if } F1 \le 0.69 \text{ or invalid output files (size and format) }  \\[1mm] 
5 & \text{if } F1 \ge 0.74 \\[1mm]
\frac{F1 - 0.69}{0.74 - 0.69} \times 5 & \text{if } 0.69 < F1 < 0.74
\end{cases}
$$

F1 Macro will be used in this part, which will be calculated using your `z{studentID}_classification.csv` output file.

The requirement may be relaxed — awarding full marks for F1 $\ge$ 0.74 — before releasing the marks for Assignment 3, if no student meets the original threshold.
